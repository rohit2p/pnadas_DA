```python
import numpy as np
import pandas as pd
```

```python
# Let's create some data that we are going to use in order to create series
d = {'a':11,'b':12,'c':13}
labels = ['a','b','c']
my_list = [22,45,66]
my_array = np.array([34,55,77])
```

```python
# creating a Seires from list
pd.Series(my_list)
```

```python
# creating a series form list but with custom index
pd.Series(my_list, index=labels)
```

```python
# creating a series from array
pd.Series(my_array)
```

```python
# creating a series from dict
data = pd.Series(d)
```
