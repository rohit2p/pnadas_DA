```python
df = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})

# Add a row using a list of values
df.loc[len(df)] = ['Charlie', 35]

# Alternatively, add a row using a dictionary (handles column mismatch safely)
df.loc[len(df)] = {'Name': 'Diana', 'Age': 28}
```

**How to update a perticular index in dataframe**

```python
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha"],
    "Math": [85, 90, 75, 88],
    "Science": [80, 95, 82, 91]
}

df = pd.DataFrame(data)

df.loc[2, "Math"] = 85
df
```

**Selecting subset of rows and columns**

```python
df.loc[[0,1]][["Math","Science"]] # this selects a subset fo row and column
```

**Conditional selecation**

```python
data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Ravi"],
    "Math": [85, 92, 76, 88, 65],
    "Science": [78, 89, 81, 95, 70]
}

df = pd.DataFrame(data)
df
```

```python
df["Math"] > 80 # boolian series
```

```python
df[df["Math"] > 80] # sutdent having more then 80 marks in maths
```

```python
df[df["Math"] == 85]
```

```python
df[df["Name"] == "Rahul"]
```

```python
df[(df["Math"] > 80) & (df["Science"] > 80)]
```

```python
df[(df["Math"] > 90) | (df["Science"] > 90)] # in pandas we dont write {and}, {or} we write {&}, {|}
```

```python
# Suppose you only want the names of students who scored above 80 in Math.
df[df["Math"] > 80]["Name"]
```

```python
# you can also write this 
df.loc[df["Math"] > 80, "Name"] # for muntiple column do this ["Name", "Math"]
```

```python
# you can create a new column based on a condition.
df["Pass"] = df["Math"] >= 40
```
