```python
import numpy as np 
import pandas as pd 
```

**Finding Missing Data**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

**Finding Missing Data**

```python
df.isna() #isna() is basically asking a question... "Is this value missing?"
```

```python
#Counting Missing Values 
df.isna().sum()
```

```python
"""
But why does .sum() count missing values?
Because Python treats:
True  = 1
False = 0
"""
```

```python
df.isna().any() # Does this column contain at least ONE missing value?
```

**Removing Missing Data**

```python
df.dropna() #Very Important: dropna() doesn't change df its simpilly drop row with NA
```

```python
df.dropna(inplace=True) # now this changes the df in the ram
```

```python
df.dropna(thresh=1) # thresh= 1, Keep rows having at least 1 non-missing value.
```

```python
df.dropna(thresh=3)
```

**Filling missing data**

```python
df.fillna(0)
```

```python
# Filling Different Columns with Different Values
values = {
    'A': 0,
    'B': 100,
    'C': 300,
    'D': 400
}
df.fillna(value=values)
```

```python
df.fillna(df.mean()) # df.mean() calculates the average of each column.
```

```python
#df["Marks"].fillna(df["Marks"].mean())
```

**You may encounter this**

```python
df.isnull() # check null value
```

```python
df.notna() #checks for values that are NOT missing.
```
